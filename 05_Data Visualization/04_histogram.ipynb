## Histogram

- It shows the distribution of numerical data by dividing values into BINS (ranges)
- It also count how many values falls into each bin (frequency)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)          # it fixes the random numbers, it generated same numbers every time
scores = np.random.normal(70,10,100)
# np.random.normal(mean, standard deviation, total values)

# bins = [0,30,70,100]
plt.hist(scores, color = "lightpink", edgecolor = "black", bins = 15)
# we can customize bins according to us by writing them in a list.

# print(scores)
```

```python
marks = [88, 74, 80, 92, 89, 60, 80, 68, 69, 74,
 71, 85, 78, 71, 74, 73, 85, 68, 73, 61,
 44, 77, 79, 63, 93, 55, 70, 68, 85, 85,
 72, 74, 61, 50, 67, 72, 82, 82, 66, 67,
 60, 56, 53, 90, 65, 66, 57, 78, 54, 68,
 61, 74, 65, 58, 70, 74, 71, 73, 64, 66,
 63, 66, 62, 53, 72, 66, 54, 75, 61, 71,
 77, 71, 81, 58, 74, 63, 61, 63, 64, 62,
 59, 71, 67, 75, 75, 72, 53, 58, 67, 72,
 60, 82, 72, 80, 74, 77, 70, 88, 71, 74]

plt.hist(marks, edgecolor = "black", bins = 10)
plt.xlabel("Marks")
plt.ylabel("No. of Students")
plt.title("Distribution of marks")
```

```python
boys_marks = [88, 74, 80, 92, 89, 60, 80, 68, 69, 74,
              71, 85, 78, 71, 74, 73, 85, 68, 73, 61]

girls_marks = [65, 70, 72, 81, 79, 68, 75, 77, 83, 69,
               71, 76, 80, 85, 74, 72, 78, 66, 70, 73]

plt.hist([boys_marks,girls_marks], bins = 5, edgecolor = "black", alpha = 0.5, label = ["Boys","Girls"])

plt.axvline(x=71, color="red", linestyle = "--", linewidth = 2, label = "Passing marks")

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")
plt.grid()
plt.legend()
plt.show()
```

axvline() is used to draw a vertical line in a graph
