### Seaborn

Seaborn is a Python data visualization library built on top of Matplotlib. It is used to create attractive and statistical graphs with less code and works very well with Pandas DataFrames.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
flights = sns.load_dataset("flights")
print(type(flights))
print(flights.head())
```

```python
tips = sns.load_dataset("tips")
print(type(tips))
print(tips.head())
```

```python
# relplot() ka matlab Relationship Plot hota hai.
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "smoker",                        # highlight smokers
    size = "size"
    # style = "smoker"
)
```

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day",
    size="size",
    style="smoker",
    col="time",
    row="sex",
    palette="Set2",                # Colors change karne ke liye.
    alpha=0.7,                     # Transparency.
    height=4,
    aspect=1.2,                    # Width adjust karta hai.
    legend="full"
)
```
