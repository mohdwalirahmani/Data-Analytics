#### Stackplots

A Stack Plot is used to show how different categories contribute to the total over time. <br>
Line chart + Filled colors + Stacked one above another

```python
import matplotlib.pyplot as plt
```

```python
# stackplots
years = [2021, 2022, 2023, 2024, 2025]

IT = [20, 25, 30, 35, 40]
HR = [10, 12, 15, 18, 20]
Sales = [15, 18, 20, 25, 30]

plt.stackplot(years, IT, HR, Sales, labels = ["IT", "HR", "Sales"])

plt.xticks(years)                          # it avoid writing years like 2021.5

plt.title("Employee in different departments")
plt.xlabel("Years")
plt.ylabel("Departments")
plt.legend()
```

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

North = [45, 50, 55, 60, 65, 75, 70]
South = [35, 38, 42, 45, 50, 55, 52]
East = [25, 28, 30, 35, 40, 45, 42]
West = [30, 32, 35, 38, 42, 48, 45]

plt.stackplot(days, North, South, East, West, labels = ["North", "South", "East", "West"])

plt.title("Weekly Customers from Different Areas")
plt.xlabel("Days")
plt.ylabel("No. of Customers")
plt.legend()
```
